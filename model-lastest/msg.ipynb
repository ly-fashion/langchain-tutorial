## 工具调用

### 1. 定义工具

```python
from langchain_core.tools import tool

@tool
def add(a: int, b: int) -> int:
    """将两个数字相加"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """将两个数字相乘"""
    return a * b

# 查看工具的 schema
print(add.name)
print(add.description)
print(add.args_schema.model_json_schema())
```

### 2. 绑定工具并调用

```python
# 将工具绑定到 LLM
llm_with_tools = llm.bind_tools([add, multiply])

# 发送消息，模型会决定是否调用工具
response = llm_with_tools.invoke("请计算 3 加 5 等于多少")

# 查看返回的 tool_calls
rprint(response)

# # 把拿到的tool_calls再次返回给llm
# tool_calls = response.tool_calls
# tool_message = ToolMessage(content=18,tool_call_id=tool_calls[0]['id'])
# messages = [HumanMessage("请计算 3 加 5 等于多少"), response,tool_message]
# final_message = llm_with_tools.invoke(messages)
# rprint(final_message)
```

### 3. 处理工具调用并获取最终结果

```python
# 定义工具映射
tools_map = {"add": add, "multiply": multiply}

# 手动执行工具调用
if response.tool_calls:
    tool_call = response.tool_calls[0]  # 取第一个工具调用
    tool_name = tool_call["name"]
    tool_args = tool_call["args"]
    
    rprint(f"[bold]调用工具:[/bold] {tool_name}")
    rprint(f"[bold]参数:[/bold] {tool_args}")
    
    # 执行工具
    tool_result = tools_map[tool_name].invoke(tool_args)
    rprint(f"[bold]结果:[/bold] {tool_result}")
    
    # 构建完整对话历史
    messages = [
        HumanMessage("请计算 3 加 5 等于多少"),
        response,  # AIMessage (包含 tool_calls)
        ToolMessage(content=str(tool_result), tool_call_id=tool_call["id"]),
    ]
    
    # 再次调用 LLM 获取最终回答
    final_response = llm_with_tools.invoke(messages)
    rprint(f"\n[bold green]最终回答:[/bold green] {final_response.content}")
else:
    rprint("[yellow]模型未调用任何工具[/yellow]")
```

### 4. 使用 `bind_tools` + 自动执行 (Tool Calling Agent)

```python
from langchain_core.tools import tool

@tool
def get_weather(city: str) -> str:
    """查询指定城市的天气信息"""
    # 模拟天气数据
    weather_data = {
        "北京": "晴天，25°C",
        "上海": "多云，22°C",
        "广州": "小雨，28°C",
    }
    return weather_data.get(city, f"暂无 {city} 的天气数据")

@tool
def calculate(expression: str) -> str:
    """计算数学表达式"""
    try:
        result = eval(expression)
        return f"计算结果: {expression} = {result}"
    except Exception as e:
        return f"计算错误: {e}"

# 绑定工具
llm_with_tools = llm.bind_tools([get_weather, calculate])

# 测试天气查询
response = llm_with_tools.invoke("今天北京和上海的天气怎么样？")
rprint(response)
```

### 5. 处理多个工具调用

```python
# 处理多个工具调用
tools_map = {"get_weather": get_weather, "calculate": calculate}

if response.tool_calls:
    messages = [
        HumanMessage("今天北京和上海的天气怎么样？"),
        response,  # AIMessage (可能包含多个 tool_calls)
    ]
    
    # 执行所有工具调用
    for tc in response.tool_calls:
        tool_name = tc["name"]
        tool_args = tc["args"]
        
        rprint(f"[bold cyan]调用工具:[/bold cyan] {tool_name}({tool_args})")
        tool_result = tools_map[tool_name].invoke(tool_args)
        
        messages.append(ToolMessage(
            content=str(tool_result),
            tool_call_id=tc["id"],
        ))
        rprint(f"[bold green]结果:[/bold green] {tool_result}\n")
    
    # 获取最终回答
    final_response = llm_with_tools.invoke(messages)
    rprint(f"[bold yellow]最终回答:[/bold yellow] {final_response.content}")
else:
    rprint("[yellow]模型未调用任何工具[/yellow]")
```

### 多模态消息

```python
import base64
import httpx
from langchain_core.messages import HumanMessage

# 下载图片并转为 base64
image_url = "https://images.unsplash.com/photo-1505740420928-5e560c06d30e?w=400"

image_data = httpx.get(image_url).content
base64_image = base64.b64encode(image_data).decode("utf-8")

# 使用 base64 格式传入图片
message = HumanMessage(
    content=[
        {"type": "text", "text": "请描述这张图片中的内容"},
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{base64_image}"
            },
        },
    ]
)

response = llm.invoke([message])
rprint(response.content)
```

#### 使用本地图片 (Base64 编码)

```python
def encode_image_to_base64(image_path: str) -> str:
    """将本地图片编码为 base64"""
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

# 使用示例（替换为你的本地图片路径）
# image_path = "path/to/your/image.jpg"
# base64_image = encode_image_to_base64(image_path)
# message = HumanMessage(
#     content=[
#         {"type": "text", "text": "请识别图片中的物品"},
#         {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"}},
#     ]
# )
# response = llm.invoke([message])
# rprint(response.content)
```

### 使用真实图片进行物品识别

```python
# 使用本地图片进行物品识别
image_url = "https://images.unsplash.com/photo-1505740420928-5e560c06d30e?w=400"

# 下载并转 base64
image_data = httpx.get(image_url).content
base64_image = base64.b64encode(image_data).decode("utf-8")

message = HumanMessage(
    content=[
        {"type": "text", "text": "请识别图片中的物品，并详细描述它的特征"},
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/jpeg;base64,{base64_image}"
            },
        },
    ]
)

response = llm.invoke([message])
rprint(f"[bold green]识别结果:[/bold green]\n{response.content}")
```

### 多图片对比识别

```python
# 多图片对比识别
image_url_1 = "https://images.unsplash.com/photo-1523275335684-37898b6baf30?w=300"
image_url_2 = "https://images.unsplash.com/photo-1526170375885-4d8ecf77b99f?w=300"

# 下载并转 base64
img1_data = base64.b64encode(httpx.get(image_url_1).content).decode("utf-8")
img2_data = base64.b64encode(httpx.get(image_url_2).content).decode("utf-8")

message = HumanMessage(
    content=[
        {"type": "text", "text": "请对比这两张图片中的物品，说明它们的区别"},
        {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{img1_data}"}},
        {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{img2_data}"}},
    ]
)

response = llm.invoke([message])
rprint(f"[bold yellow]对比结果:[/bold yellow]\n{response.content}")
```

### 流式输出识别结果

```python
# 流式输出图片识别结果
image_url = "https://images.unsplash.com/photo-1505740420928-5e560c06d30e?w=400"
image_data = base64.b64encode(httpx.get(image_url).content).decode("utf-8")

message = HumanMessage(
    content=[
        {"type": "text", "text": "用中文详细介绍这张图片"},
        {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{image_data}"}},
    ]
)

# 流式输出
print("识别结果：", end="")
for chunk in llm.stream([message]):
    print(chunk.content, end="", flush=True)
print()
```
